```python
import sys
import os
import time
import pandas as pd

# Asegurar que Python encuentra los archivos en la carpeta src
sys.path.append(os.path.abspath('./src'))

from problem import CampusGuardProblem
from search import bfs, dfs, a_star
```

```python
# Definir Estado Inicial (Alto Riesgo, 35% Inasistencias, Sin Acción)
s0 = ('Alto', 35, 'Sin_Accion')
problem = CampusGuardProblem(s0)
print(f"Estado Inicial: {s0}")
```

```python
# Ejecutar Algoritmos y medir tiempos
algoritmos = {
    "BFS": bfs,
    "DFS": dfs,
    "A*": a_star
}

resultados = []

for nombre, funcion in algoritmos.items():
    start = time.perf_counter()
    nodo = funcion(problem)
    end = time.perf_counter()
    
    tiempo = round((end - start) * 1000, 2)
    camino = nodo.path() if nodo else []
    costo = nodo.path_cost if nodo else -1
    
    resultados.append({
        "Algoritmo": nombre,
        "Tiempo (ms)": tiempo,
        "Costo g(n)": costo,
        "Camino de Intervención": " -> ".join(camino)
    })

# Mostrar Tabla Comparativa
df = pd.DataFrame(resultados)
display(df)
```
