### Celda 3: Punto B (CampusGuard UMNG) — Carga Inicial de Políticas Normativas en la KB
Integramos las proposiciones de la actividad de clase (`Promedio_Bajo`, `Ausentismo_Alto`, `Pago_Matricula_Pendiente`, `Plazo_Vencido`) junto con la continuidad de los Cortes 1 y 2 (`Materias_Criticas_Mayor_2`), registrando **4 reglas condicionales** en la capa declarativa mediante `tell_regla()`.

```python
# 1. Instanciar la Base de Conocimiento declarativa y el Agente CampusGuard
kb_campus = BaseDeConocimiento()
agente_campus = AgenteBasadoEnConocimiento(kb_campus)

# 2. Registro de políticas normativas institucionales (Capa Declarativa - 4 Reglas)
kb_campus.tell_regla(
    premisas=["Promedio_Bajo", "Ausentismo_Alto"],
    consecuencia="Alerta_Riesgo_Academico",
    descripcion="Bajo rendimiento y alto ausentismo (> 40%) disparan alerta temprana (< 48h)."
)
kb_campus.tell_regla(
    premisas=["Alerta_Riesgo_Academico"],
    consecuencia="Activar_Tutor_Acompanamiento",
    descripcion="Toda alerta de riesgo académico obliga a asignar tutor pedagógico y remitir a Bienestar."
)
kb_campus.tell_regla(
    premisas=["Alerta_Riesgo_Academico", "Materias_Criticas_Mayor_2"],
    consecuencia="Requiere_Reestructurar_Carga",
    descripcion="Alerta académica con >= 2 materias críticas habilita reestructuración de créditos."
)
kb_campus.tell_regla(
    premisas=["Pago_Matricula_Pendiente", "Plazo_Vencido"],
    consecuencia="Bloqueo_Plataforma_LMS",
    descripcion="Mora financiera tras vencimiento de plazo genera restricción administrativa en LMS."
)

# 3. Catálogo de intervenciones disponibles para el agente (orden de prioridad institucional)
acciones_campusguard = {
    "Asignar_Tutor_Pedagogico": "Activar_Tutor_Acompanamiento",
    "Reestructurar_Carga_Academica": "Requiere_Reestructurar_Carga",
    "Bloquear_Inscripcion_LMS": "Bloqueo_Plataforma_LMS"
}

print('\nBase de Conocimiento de CampusGuard inicializada con 4 reglas normativas.')
```

### Celda 4: Simulación Escenario t = 0 (Inicio de Semestre — Monitoreo Preventivo)
El estudiante presenta `Promedio_Bajo` y `Pago_Matricula_Pendiente` (plazo ordinario vigente), pero su ausentismo, vencimiento de plazo y materias críticas están controlados (`False`). Al consultar mediante `ASK`, no se activa ninguna alerta crítica.

```python
perceptos_t0 = {
    "Promedio_Bajo": True,
    "Ausentismo_Alto": False,
    "Pago_Matricula_Pendiente": True,
    "Plazo_Vencido": False,
    "Materias_Criticas_Mayor_2": False
}

agente_campus.percibir(perceptos_t0)
agente_campus.decidir_accion(acciones_campusguard)
```

### Celda 5: Simulación Escenario t = 1 (Alerta por Ausentismo > 40% -> Tutoría Obligatoria)
En el instante $t = 1$, ingresa vía `TELL` el reporte `Ausentismo_Alto = True`. El motor de inferencia aplica **Modus Ponens** en cadena (`Alerta_Riesgo_Academico` $\rightarrow$ `Activar_Tutor_Acompanamiento`) y autoriza `Asignar_Tutor_Pedagogico`.

```python
perceptos_t1 = {
    "Ausentismo_Alto": True
}

agente_campus.percibir(perceptos_t1)
agente_campus.decidir_accion(acciones_campusguard)
```

### Celda 6: Simulación Escenario t = 2 (Corte Académico -> Reestructuración de Carga)
En el instante $t = 2$ (continuidad con Cortes 1 y 2), ingresa `Materias_Criticas_Mayor_2 = True` mientras el estudiante aún conserva acceso activo al LMS. El motor deduce `Requiere_Reestructurar_Carga` y ejecuta `Reestructurar_Carga_Academica`.

```python
perceptos_t2 = {
    "Materias_Criticas_Mayor_2": True
}

agente_campus.percibir(perceptos_t2)
agente_campus.decidir_accion(acciones_campusguard)
```

### Celda 7: Simulación Escenario t = 3 (Cierre Plazo Financiero -> Bloqueo LMS y Auditoría KB)
En el instante $t = 3$, vence el plazo de matrícula (`Plazo_Vencido = True`). Gracias a la **memoria persistente** de la Base de Conocimiento (`Pago_Matricula_Pendiente` registrado desde $t=0$), el motor infiere `Bloqueo_Plataforma_LMS`, ejecuta `Bloquear_Inscripcion_LMS` y presenta la auditoría completa de la KB.

```python
perceptos_t3 = {
    "Plazo_Vencido": True
}

agente_campus.percibir(perceptos_t3)
agente_campus.decidir_accion(acciones_campusguard)

# Verificación directa ASK sobre la restricción en plataforma
autoriza_bloqueo = kb_campus.ask('Bloqueo_Plataforma_LMS')
print(f"\nVerificación directa ASK('Bloqueo_Plataforma_LMS'): {autoriza_bloqueo}")

# Auditoría transparente de todos los hechos verdaderos almacenados e inferidos en la KB
print("\n===========================================================")
print(" AUDITORÍA FINAL DE HECHOS EN LA BASE DE CONOCIMIENTO (KB)")
print("===========================================================")
for hecho in sorted(kb_campus.hechos):
    print(f"  • {hecho}")
```
